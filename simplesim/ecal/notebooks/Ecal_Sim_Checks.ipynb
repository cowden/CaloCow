```python
import pylab as plt
import h5py
import numpy as np
```

```python
import os
```

```python
os.listdir("/data/me/calox/dkr/ecal/")
```

```python
f20 = h5py.File("/data/me/calox/dkr/ecal/simple-ecal-20GeV.h5")
X20 = np.array([f20["data"][k] for k in f20["data"].keys()])
```

```python
plt.hist(X20[:,0]/1000, bins="fd");
plt.xlabel(r"$E_{beam}$ [GeV]");
plt.ylabel("Count");
```

```python
plt.scatter(X20[:,1], X20[:,2])
```

```python
plt.matshow(X20[:,3:].mean(axis=0).reshape((5,5)));
plt.colorbar();
```

```python
plt.hist(X20[:,3:].sum(axis=1)/1000, bins="fd");
plt.xlabel(r"$\hat{E}$ [GeV]");
plt.ylabel("Count");
```

```python
plt.scatter(X20[:,0]/1000, X20[:,3:].sum(axis=1)/1000)
plt.xlabel(r"$E_{beam}$ [GeV]");
plt.ylabel(r"$\hat{E}$ [GeV]");
```

```python
f30 = h5py.File("/data/me/calox/dkr/ecal/simple-ecal-30GeV.h5","r")
X30 = np.array([f30["data"][k] for k in f30["data"].keys()])
```

```python
plt.scatter(
    np.concatenate((X20[:,0]/1000, X30[:,0]/1000)),
    np.concatenate((X20[:,3:].sum(axis=1)/1000, X30[:,3:].sum(axis=1)/1000))
)
```

```python
f50 = h5py.File("/data/me/calox/dkr/ecal/simple-ecal-50GeV.h5","r")
X50 = np.array([f50["data"][k] for k in f50["data"].keys()])
```

```python
plt.scatter(
    np.concatenate((X20[:,0]/1000, X30[:,0]/1000, X50[:,0]/1000)),
    np.concatenate((X20[:,3:].sum(axis=1)/1000, X30[:,3:].sum(axis=1)/1000, X50[:,3:].sum(axis=1)/1000))
)
```
